## Importar Datos

```python
import pandas as pd
pd.set_option('display.float_format', lambda x: f'{x:,.2f}')

# 1) Cargar dataset
data = pd.read_csv('cac_ltv_model.csv')
```

### 🧮 Fórmulas de negocio (a usar en el examen)
- **CAC por canal**:  
    $$ CAC = \frac{\text{Total Marketing Spend}}{\text{Clientes adquiridos}} $$
  
- **LTV por observación** (aprox.):  
   $$ LTV = ARPU \times contract\_length\_months \times gross\_margin \times (1 - churn\_rate) $$

- **Unit economics**:  
  $$ LTV/CAC   $$(ideal \> 3 como regla general de SaaS)

## 📊 Parte B – Agrupaciones (SOLO agrupar)
4. **Agrupa** por `acquisition_channel` y calcula **ARPU promedio**. Ordena de mayor a menor.  


```python
arpu_prom_channel = data.groupby('acquisition_channel')['arpu'].mean().sort_values(ascending=False)
print(arpu_prom_channel)
```

5. **Agrupa** por `region` y calcula **churn_rate promedio**. Identifica la región con mayor churn.  



```python
churn_rate_prom_region = data.groupby('region')['churn_rate'].mean().sort_values(ascending=False)
print("churn_rate promedio por region:")
print(churn_rate_prom_region)
```

6. **Agrupa** por `customer_tier` y calcula **marketing_spend total**. ¿Cuál tier consume más presupuesto?

```python
marketing_spend_total = data.groupby('customer_tier')['marketing_spend'].sum()
print("marketing_spend total por tier:")
print(marketing_spend_total)
```

## 🔀 Parte C – Filtro **y** Agrupación (Ambas)
7. **Filtra** solo registros de **`signup_source = 'web'`** y luego **agrupa** por `acquisition_channel` para obtener el **churn_rate promedio**.  


```python
web_users = data[data['signup_source'] == 'web']
churn_web_channel = web_users.groupby('acquisition_channel')['churn_rate'].mean()

print("Churn promedio por canal")
print(churn_web_channel)
```

8. **Filtra** solo **`mobile_app`** y luego **agrupa** por `region` para calcular **ARPU promedio**.  


```python
mobile_app_users = data[data['signup_source'] == 'mobile_app']
arpu_mobile_region = mobile_app_users.groupby('region')['arpu'].mean().sort_values(ascending=False)
print("ARPU promedio por region")
print(arpu_mobile_region)
```

9. **Filtra** a clientes **`Pro`** y **agrupa** por `acquisition_channel` para calcular **marketing_spend total**.


```python
acquis_channel_pro = data[data['customer_tier'] == 'Pro']
marketing_spend_pro = acquis_channel_pro.groupby('acquisition_channel')['marketing_spend'].sum()
print("marketing_spend total por canal")
print(marketing_spend_pro)
```

## 📈 Parte D – Métricas de negocio (CAC, LTV, LTV/CAC)
10. **CAC por canal (Agrupa)**: calcula el CAC de cada `acquisition_channel` como:  
   `CAC = marketing_spend_total_del_canal / #clientes_del_canal`  


```python
cac_channel = (data.groupby('acquisition_channel')['marketing_spend'].sum() / data.groupby('acquisition_channel').size())
print("CAC por canal:")
print(cac_channel.sort_values(ascending=False))
```

11. **CAC por tier (Agrupa)**: calcula el CAC de cada `customer_tier` como:  
   `CAC = marketing_spend_total_del_tier / #clientes_del_canal`  


```python
cac_tier = (data.groupby('customer_tier')['marketing_spend'].sum() / data.groupby('customer_tier').size())
print("CAC por tier:")
print(cac_tier.sort_values(ascending=False))
```

12. **LTV por tier (Ambas)**: crea una columna `ltv_individual` con la fórmula dada.


```python
ltv_individual = (data['arpu'] * data['contract_length_months'] * data['gross_margin'] * (1 - data['churn_rate']))
data['ltv_individual'] = ltv_individual
data.head(5)
```

12. A. Luego **agrupa** `ltv_individual` por
`acquisition_channel` para obtener el **LTV promedio**.  

```python
groupby_channel = data.groupby('acquisition_channel')['ltv_individual'].mean().sort_values(ascending=False)
print("LTV promedio por acquisition_channel:")
print(groupby_channel)
```

12. B. Luego **agrupa** `ltv_individual`por `customer_tier` para obtener el **LTV promedio**.





```python
LTV = data.groupby('customer_tier')['ltv_individual'].mean().sort_values(ascending=False)

print("LTV promedio por canal:")
print(LTV)
```

13. **Unit economics (Ambas)**: combina tus resultados para comparar **LTV promedio por tier** contra **CAC por canal** y comenta **qué combinaciones canal + tier** lucen más saludables (busca **LTV/CAC > 3**).

```python
import pandas as pd

# 1) Definir los LTV promedio por tier (de tus cálculos anteriores)
ltv_tier = {
    "Enterprise": 1349.02,
    "Pro": 577.63,
    "Basic": 291.77
}

# 2) Definir los CAC por canal (de tus cálculos anteriores)
cac_channel = {
    "google_ads": 401.02,
    "meta_ads": 349.27,
    "outbound_sales": 200.22,
    "organic_search": 49.95
}

# 3) Convertirlos a DataFrames
ltv_df = pd.DataFrame.from_dict(ltv_tier, orient="index", columns=["LTV"])
cac_df = pd.DataFrame.from_dict(cac_channel, orient="index", columns=["CAC"])

# 4) Calcular la matriz de LTV/CAC (cruce tier × canal)
results = pd.DataFrame(
    {
        canal: ltv_df["LTV"] / cac
        for canal, cac in cac_channel.items()
    }
)

# 5) Redondear a 2 decimales para visualización
results = results.round(2)

print("Matriz LTV/CAC (Tier x Canal):")
print(results)
```
